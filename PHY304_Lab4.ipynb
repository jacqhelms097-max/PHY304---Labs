Analytical definitions of air drag:

1) Linear
 
ma = -mg - bv


@terminal velocity the acceleration must be 0

0 = -mg - bv

v = -mg/b


2) Quadratic
   
ma = -mg - cv|v|

0 = -mg - cv|v|

v = -sqrt(mg/c)




```python
import numpy as np
import matplotlib.pyplot as plt

def euler_newton(force, times, x0, v0, m):
    nsteps = len(times)
    x = np.zeros(nsteps)
    v = np.zeros(nsteps)
    x[0] = x0
    v[0] = v0
  
    
    for i in range(nsteps - 1):

        dt = times[i + 1] - times[i] 
        a = force(times[i], x[i], v[i]) / m
        v[i + 1] = v[i] + a * dt
        x[i + 1] = x[i] + v[i + 1] * dt
        
    return x, v

def linear_drag(t, x, v):
    return -m*g - b*v

g = 9.8
b = 0.5
m = 1
v0 = 0
x0 = 300
 
times = np.linspace(0, 20, 1000)
 
x, v = euler_newton(linear_drag, times, x0, v0, m)
 
terminal_velocity_L = -m*g/b

    
plt.figure()
 
plt.plot(times, x)
plt.xlabel("Time (s)")
plt.ylabel("Position (m)")
plt.title("Linear Drag (Position vs. Time)")
plt.show()

 

plt.plot(times, v, label="Numerical Velocity")
plt.axhline(terminal_velocity_L, linestyle="--", label="Terminal Velocity")
plt.xlabel("Time (s)")
plt.ylabel("Velocity (m/s)")
plt.title("Linear Drag (Velocity vs. Time)")
plt.legend()
plt.show()
```

```python
c = 0.2
def quadratic_drag(t, x, v):
    return -m*g - c*v*abs(v)

x2, v2 = euler_newton(quadratic_drag, times, x0, v0, m)
terminal_velocity_Q = -np.sqrt(m*g/c)

plt.figure()

plt.plot(times, x2)
plt.xlabel("Time (s)")
plt.ylabel("Position (m)")
plt.title("Quadratic Drag (Position vs. Time)")
plt.show()


plt.plot(times, v2, label="Numerical Velocity")
plt.axhline(terminal_velocity_Q, linestyle="--", label="Terminal Velocity")
plt.xlabel("Time (s)")
plt.ylabel("Velocity (m/s)")
plt.title("Quadratic Drag (Velocity vs Time)")
plt.legend()
plt.show()
```

Both plots reach the constant calculated terminal velocity, but the quadratic graph does so at a much faster rate. Linear terminal velocity seems to be at around 11 or 12 seconds, while quadratic terminal velocity is at only around 2.5 seconds.

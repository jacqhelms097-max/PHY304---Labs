```python
import numpy as np
import matplotlib.pyplot as plt

def euler_newton(force, times, x0, v0, m):
    nsteps = len(times)
    x = np.zeros(nsteps)
    v = np.zeros(nsteps)
    x[0] = x0
    v[0] = v0
  
    
    for i in range(nsteps - 1):

        dt = times[i + 1] - times[i] 
        a = force(times[i], x[i], v[i]) / m
        v[i + 1] = v[i] + a * dt
        x[i + 1] = x[i] + v[i + 1] * dt
        
    return x, v

def gravity(t, x, v):   
    return -9.8
x0 = 10
v0 = 0
m = 1

times = np.linspace(0, 3, 100)

x, v = euler_newton(gravity, times, x0, v0, m)
g = 9.8
    
position = x0 + v0*times - 0.5*g*times**2
velocity = v0 - g*times

    
plt.figure()
 
plt.plot(times, x, label="Numerical")
plt.plot(times, position, "--", label="Analytical")
plt.xlabel("Time (s)")
plt.ylabel("Position (m)")
plt.legend()
plt.title("Position vs Time")
plt.show()

 

plt.plot(times, v, label="Numerical")
plt.plot(times, velocity, "--", label="Analytical")
plt.xlabel("Time (s)")
plt.ylabel("Velocity (m/s)")
plt.title("Velocity vs Time")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def euler_newton(force, times, x0, v0, m):
    nsteps = len(times)
    x = np.zeros(nsteps)
    v = np.zeros(nsteps)
    x[0] = x0
    v[0] = v0
  
    
    for i in range(nsteps - 1):

        dt = times[i + 1] - times[i] 
        a = force(times[i], x[i], v[i]) / m
        v[i + 1] = v[i] + a * dt
        x[i + 1] = x[i] + v[i + 1] * dt
        
    return x, v

def spring(t, x, v):
    k = 1
    return -k * x

x0 = 1
v0 = 0
m = 1
 
times = np.linspace(0, 20, 1000)
 
x, v = euler_newton(spring, times, x0, v0, m)
 
position = np.cos(times)
velocity = -np.sin(times)

    
plt.figure()
 
plt.plot(times, x, label="Numerical")
plt.plot(times, position, "--", label="Analytical")
plt.xlabel("Time (s)")
plt.ylabel("Position (m)")
plt.legend()
plt.title("Simple Harmonic Oscillator (Position)")
plt.show()

 

plt.plot(times, v, label="Numerical")
plt.plot(times, velocity, "--", label="Analytical")
plt.xlabel("Time (s)")
plt.ylabel("Velocity (m/s)")
plt.title("Simple Harmonic Oscillator (Velocity)")
plt.legend()
plt.show()
```
